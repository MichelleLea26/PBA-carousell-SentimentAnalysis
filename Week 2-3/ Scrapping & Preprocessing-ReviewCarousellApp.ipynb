```python
import matplotlib.pyplot as plt
import seaborn as sns
import regex
import re
import string
import emoji
from collections import Counter
from textblob import TextBlob

import nltk
from nltk import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

import wordcloud
from wordcloud import WordCloud, STOPWORDS

from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
```

**Case Folding & Cleansing**

```python
print("Memulai proses pembersihan teks...")

# Menggunakan data dari hasil scraping sebelumnya (df_carousell_clean)
df_prep = df_carousell_fil.copy()

```

```python
# Ubah ke huruf kecil (lowercase)
df_prep['content'] = df_prep['content'].astype(str)
df_prep['cleaned_content'] = df_prep['content'].str.lower()

```

```python
# Fungsi untuk membersihkan angka, emoji, dan tanda baca
def clean_text(text):
    text = re.sub(r'[\U00010000-\U0010ffff]', '', text) # Hapus emoji
    text = re.sub(r'\d+', '', text)                      # Hapus angka
    text = re.sub(r'[^a-z\s]', '', text)                 # Hapus tanda baca (sisakan huruf dan spasi)
    return text

df_prep['cleaned_content'] = df_prep['cleaned_content'].apply(clean_text)
```

Normalisasi Slang

```python
# Kamus normalisasi kata tidak baku / slang Indonesia
slang_dict = {
    'gk': 'tidak', 'ga': 'tidak', 'gak': 'tidak', 'nggak': 'tidak',
    'ngga': 'tidak', 'tdk': 'tidak', 'tp': 'tapi', 'krn': 'karena',
    'karna': 'karena', 'udh': 'sudah', 'udah': 'sudah', 'sdh': 'sudah',
    'blm': 'belum', 'blum': 'belum', 'msh': 'masih', 'org': 'orang',
    'yg': 'yang', 'dg': 'dengan', 'dgn': 'dengan', 'jg': 'juga',
    'juga': 'juga', 'utk': 'untuk', 'bgt': 'sangat', 'banget': 'sangat',
    'sy': 'saya', 'aku': 'saya', 'gue': 'saya', 'gw': 'saya',
    'kmu': 'kamu', 'lo': 'kamu', 'lu': 'kamu', 'lg': 'lagi',
    'lgi': 'lagi', 'bisa': 'bisa', 'bs': 'bisa', 'app': 'aplikasi',
    'apk': 'aplikasi', 'update': 'perbarui', 'error': 'eror',
    'loading': 'memuat', 'chat': 'pesan', 'seller': 'penjual',
    'buyer': 'pembeli', 'scam': 'penipuan', 'banned': 'diblokir',
    'lemot': 'lambat', 'lelet': 'lambat', 'mantap': 'bagus',
    'mantul': 'bagus', 'keren': 'bagus', 'oke': 'baik', 'ok': 'baik',
    'bagus': 'bagus', 'jelek': 'buruk', 'parah': 'buruk',
    'smg': 'semoga', 'mksh': 'terima kasih', 'makasih': 'terima kasih',
    'thx': 'terima kasih', 'thanks': 'terima kasih',
}

def normalize_slang(text):
    words = text.split()
    return ' '.join([slang_dict.get(w, w) for w in words])

df_prep['normalized_content'] = df_prep['cleaned_content'].apply(normalize_slang)
print("Selesai: normalisasi slang")
```

Stopword Removal

```python
# Gabungkan stopword dari Sastrawi dan NLTK Bahasa Indonesia
sw_factory = StopWordRemoverFactory()
sastrawi_stopwords = set(sw_factory.get_stop_words())

nltk_stopwords = set(stopwords.words('indonesian'))

# Custom kata tidak informatif yang spesifik untuk konteks review
custom_stopwords = {
    'aplikasi', 'app', 'carousell', 'star', 'bintang',
    'play', 'store', 'google', 'android',
}

all_stopwords = sastrawi_stopwords | nltk_stopwords | custom_stopwords

def remove_stopwords(text):
    tokens = text.split()
    return ' '.join([w for w in tokens if w not in all_stopwords and len(w) > 2])

df_prep['no_stopword_content'] = df_prep['normalized_content'].apply(remove_stopwords)
print(f"Stopword dihapus. Contoh:\n{df_prep['no_stopword_content'].iloc[0]}")
```

Stemming

```python
# Inisialisasi stemmer Sastrawi
factory = StemmerFactory()
stemmer = factory.create_stemmer()

def stem_text(text):
    return stemmer.stem(text)

# Proses stemming
df_prep['stemmed_content'] = df_prep['no_stopword_content'].apply(stem_text)
print("Selesai: stemming Sastrawi")
df_prep[['content', 'cleaned_content', 'normalized_content',
         'no_stopword_content', 'stemmed_content']].head(3)
```

```python
# Simpan hasil preprocessing lengkap
df_prep.to_csv('carousell_preprocessed.csv', index=False)
print("Data preprocessing tersimpan: 'carousell_preprocessed.csv'")
print(f"Shape akhir: {df_prep.shape}")

# Cek pipeline preprocessing
sample = df_prep.sample(1)
print("\n[Asli]     :", sample['content'].values[0])
print("[Bersih]   :", sample['cleaned_content'].values[0])
print("[Normalisasi]:", sample['normalized_content'].values[0])
print("[No Stopword]:", sample['no_stopword_content'].values[0])
print("[Stemmed]  :", sample['stemmed_content'].values[0])
```
